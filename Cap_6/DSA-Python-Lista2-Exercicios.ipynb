#### Exercício 1

Escreva uma função que receba os valores de três lados de um triângulo e o classifique como "Equilátero" (todos os lados iguais), "Isósceles" (dois lados iguais) ou "Escaleno" (todos os lados diferentes).

```python
# Solução

ladoa= input(" primeiro lado do triangulo")
ladob= input(" segundo lado do triangulo")
ladoc= input(" terceiro lado do triangulo")

def classificador_Tri(ladoa, ladob, ladoc):
    if ladoa == ladob ==ladoc:
        return "Equilátero"
    if ladoa== ladob or ladoa== ladoc or ladob==ladoc:
        return "Isóceles"
    else:
        return "Escaleno"

print(classificador_Tri(ladoa, ladob, ladoc))
```

#### Exercício 3 

Você recebeu um dicionário com os nomes dos alunos e suas respectivas notas. Escreva uma função que calcula a média da turma e retorna uma lista com os nomes dos alunos que tiveram nota acima da média.

```python
# Solução
dict= { "celio": 32, "deodato": 35, "manoel":23, "daniel": 12, "enzho": 98, "pedro": 55}

def alunos_media(dict):

    #verifica se o dicionario está vazio
    if not dict:
        return[]

    # média
    soma_n= sum(dict.values())
    media= soma_n/len(dict)

    #acima da media 
    acima_media=[]
    for aluno,nota in dict.items():
        if nota> media:
            acima_media.append(aluno)
    return acima_media

resultado= alunos_media(dict)

soma_n= sum(dict.values())
media= soma_n/len(dict)

print(f"notas {dict}")
print(f"media {media}")
print(f" alunos acima da media: {resultado}")
        
        
```

#### Exercício 6

Você tem uma lista de dicionários, onde cada dicionário representa uma pessoa com nome e idade. Use a função sorted() com uma expressão lambda como chave (key) para ordenar a lista de pessoas da mais nova para a mais velha.

```python
# Solução
pessoas = [
    {"nome": "celio", "idade": 27},
    {"nome": "manoel", "idade": 43},
    {"nome": "deodato", "idade": 46},
    {"nome": "otavio", "idade": 15},
    {"nome": "ana", "idade": 66}
]

pessoas_ordenadas= sorted(pessoas,key=lambda pessoa:pessoa["idade"])

for pessoa in pessoas_ordenadas:
    print(f"{pessoa["nome"]}: {pessoa["idade"]} anos ")
```

Crie uma função que receba uma lista de números inteiros e retorne um dicionário contendo a contagem de quantos números são pares e quantos são ímpares.

```python
# Solução
lista=[ 1,33,56,2,667,43,46,12,25,76,42,55, 45]



def funcao(lista):
    contagem={"pares": 0, "impares":0}
    
    for numero in lista:
        if numero %2 ==0:
            contagem["pares"] +=1
        else:
            contagem["impares"] +=1
    return contagem

print(funcao(lista))
    
    
```

#### Exercício 8

Crie uma função que receba uma lista de strings (potenciais e-mails) e um parâmetro opcional dominio_desejado (com valor padrão "gmail.com"). A função deve retornar uma nova lista, usando list comprehension, contendo apenas os e-mails que terminam com o domínio desejado.

```python
def filtrar_emails_validos(lista_emails, dominio_desejado="gmail.com"):
    """
    Versão com validação adicional para garantir que seja um formato de e-mail.
    """
    # Filtra apenas strings e que contenham "@" (formato básico de e-mail)
    emails_validos = [
        email for email in lista_emails 
        if isinstance(email, str) and "@" in email and email.endswith(f"@{dominio_desejado}")
    ]
    
    return emails_validos

emails_mistos = [
    "usuario@gmail.com",
    "email_invalido",  # Sem @
    123,  # Não é string
    "outro@gmail.com",
    None,  # None
    "teste@hotmail.com"
]

print(filtrar_emails_validos(emails_mistos))
```

#### Exercício 9

Dada uma lista de frases, use a função map() em conjunto com uma expressão lambda para criar uma nova lista onde cada frase é convertida para letras maiúsculas e tem a palavra "PYTHON" anexada ao final.

```python
# Solução
lista = [ "eu te amo", " macacheira", " pepino", "asafe", " flor", " azazel", " branco "," o mundo vai ser melhor"]

nova_lista= list(map(lambda frase: frase.strip().upper() + " PYTHON", lista))
# "Para cada frase: remova espaços extras, converta para maiúsculas e adicione ' PYTHON' no final"

for frase in nova_lista:
    print(frase)
```
